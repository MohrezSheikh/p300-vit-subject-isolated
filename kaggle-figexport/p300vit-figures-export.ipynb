# Figure Export

Copies only `figures/*` from `p300vit-figures` (skips the dataset cache) and
prints the provenance JSON. CPU, seconds.


```python
import glob, json, os, shutil

src = None
for root, dirs, files in os.walk("/kaggle/input"):
    if "fig7_xdawn.png" in files:
        src = root
        break
assert src, "figures not found among inputs: " + str(sorted(glob.glob("/kaggle/input/*")))
os.makedirs("/kaggle/working/figures", exist_ok=True)
for f in glob.glob(os.path.join(src, "*")):
    shutil.copy2(f, "/kaggle/working/figures/")
    print("copied", os.path.basename(f))
with open("/kaggle/working/figures/example_provenance.json") as f:
    print(json.dumps(json.load(f), indent=2))
```
