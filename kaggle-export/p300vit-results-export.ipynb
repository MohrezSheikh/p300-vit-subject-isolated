# P300-ViT Results Export

Re-publishes **only the result artifacts** of `p300vit-standalone` (skip the
multi-GB preprocessed-epoch cache): all JSON/NPZ/PNG under
`results/xdawn_vit/`. Mounts the completed kernel via kernel_sources; no GPU,
no data attachments, finishes in seconds.


```python
import glob, json, os, shutil

# recursive search: kernel outputs may mount several levels deep
marker = None
for root, dirs, files in os.walk("/kaggle/input"):
    if "xdawn_vit_summary.json" in files:
        marker = root   # .../results/xdawn_vit
        break
assert marker, ("standalone results not found; mounted:", 
                sorted(glob.glob("/kaggle/input/*") + glob.glob("/kaggle/input/*/*"))[:20])
print("results dir:", marker)

DST = "/kaggle/working/results/xdawn_vit"
os.makedirs(DST, exist_ok=True)
n = 0
for path in glob.glob(os.path.join(marker, "**", "*"), recursive=True):
    if os.path.isdir(path) or path.endswith((".pt", ".yaml")):
        continue
    rel = os.path.relpath(path, marker)
    dst = os.path.join(DST, rel)
    os.makedirs(os.path.dirname(dst), exist_ok=True)
    shutil.copy2(path, dst)
    n += 1
print(f"copied {n} artifacts -> {DST}")

with open(os.path.join(DST, "xdawn_vit_summary.json")) as f:
    print(json.dumps(json.load(f), indent=2))
```
